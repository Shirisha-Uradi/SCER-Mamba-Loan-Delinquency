```python
import torch
import torch.nn as nn
from transformers import MambaConfig, MambaModel


class SCERMamba(nn.Module):

    def __init__(
        self,
        temporal_features=7,
        static_features=12,
        hidden_size=32
    ):
        super().__init__()

        # -----------------------------------
        # 1. Static borrower encoder
        # -----------------------------------
        self.static_encoder = nn.Sequential(
            nn.Linear(static_features, 32),
            nn.ReLU(),
            nn.Dropout(0.20),

            nn.Linear(32, hidden_size),
            nn.ReLU()
        )

        # -----------------------------------
        # 2. Temporal feature projection
        # -----------------------------------
        self.temporal_projection = nn.Linear(
            temporal_features,
            hidden_size
        )

        # -----------------------------------
        # 3. Static conditioning
        # -----------------------------------
        self.gamma = nn.Linear(hidden_size, hidden_size)
        self.beta = nn.Linear(hidden_size, hidden_size)

        # -----------------------------------
        # 4. REAL Mamba temporal encoder
        # -----------------------------------
        mamba_config = MambaConfig(
            vocab_size=2,
            hidden_size=hidden_size,
            state_size=16,
            num_hidden_layers=1,
            expand=2,
            conv_kernel=4,
            use_cache=False
        )

        self.mamba = MambaModel(mamba_config)

        # -----------------------------------
        # 5. Risk-event attention
        # -----------------------------------
        self.temporal_attention = nn.Linear(
            hidden_size,
            1
        )

        # 30+, 60+, 90+ delinquency event signals
        self.risk_event_attention = nn.Linear(
            3,
            1,
            bias=False
        )

        # -----------------------------------
        # 6. Adaptive gated fusion
        # -----------------------------------
        self.fusion_gate = nn.Linear(
            hidden_size * 2,
            hidden_size
        )

        # -----------------------------------
        # 7. Final risk prediction
        # -----------------------------------
        self.classifier = nn.Sequential(
            nn.Linear(hidden_size, 16),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(16, 1)
        )


    def forward(self, sequence, static):

        # ==================================================
        # Static borrower representation
        # ==================================================

        static_embedding = self.static_encoder(static)

        # ==================================================
        # Convert 7 temporal features -> hidden dimension
        # ==================================================

        temporal = self.temporal_projection(sequence)

        # ==================================================
        # Static-conditioned temporal sequence
        # ==================================================

        gamma = torch.tanh(
            self.gamma(static_embedding)
        ).unsqueeze(1)

        beta = self.beta(
            static_embedding
        ).unsqueeze(1)

        conditioned_sequence = (
            temporal * (1 + gamma)
        ) + beta

        # ==================================================
        # MAMBA
        # ==================================================

        mamba_output = self.mamba(
            inputs_embeds=conditioned_sequence,
            use_cache=False
        ).last_hidden_state

        # ==================================================
        # Risk-event attention
        #
        # sequence feature positions:
        # 4 = is_30plus_delinquent
        # 5 = is_60plus_delinquent
        # 6 = is_90plus_delinquent
        # ==================================================

        risk_events = sequence[:, :, 4:7]

        temporal_scores = self.temporal_attention(
            mamba_output
        ).squeeze(-1)

        event_scores = self.risk_event_attention(
            risk_events
        ).squeeze(-1)

        attention_scores = (
            temporal_scores + event_scores
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=1
        )

        temporal_embedding = torch.sum(
            mamba_output *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        # ==================================================
        # Adaptive gated fusion
        # ==================================================

        fusion_input = torch.cat(
            [
                temporal_embedding,
                static_embedding
            ],
            dim=1
        )

        gate = torch.sigmoid(
            self.fusion_gate(fusion_input)
        )

        fused_embedding = (
            gate * temporal_embedding
            +
            (1 - gate) * static_embedding
        )

        # ==================================================
        # Final prediction
        # ==================================================

        logits = self.classifier(
            fused_embedding
        ).squeeze(-1)

        return logits, attention_weights
```

```python
import torch
from torch.utils.data import TensorDataset, DataLoader

train_dataset = TensorDataset(
    torch.tensor(X_seq_train, dtype=torch.float32),
    torch.tensor(X_static_train, dtype=torch.float32),
    torch.tensor(y_train, dtype=torch.float32)
)

val_dataset = TensorDataset(
    torch.tensor(X_seq_val, dtype=torch.float32),
    torch.tensor(X_static_val, dtype=torch.float32),
    torch.tensor(y_val, dtype=torch.float32)
)

test_dataset = TensorDataset(
    torch.tensor(X_seq_test, dtype=torch.float32),
    torch.tensor(X_static_test, dtype=torch.float32),
    torch.tensor(y_test, dtype=torch.float32)
)

train_loader = DataLoader(
    train_dataset,
    batch_size=512,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=1024,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=1024,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("DataLoaders created successfully")
print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

```python
import torch
import torch.nn as nn

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)

model = SCERMamba().to(device)

num_positive = float(y_train.sum())
num_negative = float((y_train == 0).sum())

pos_weight_value = num_negative / num_positive

print("Risky:", int(num_positive))
print("Non-risky:", int(num_negative))
print("Positive class weight:", pos_weight_value)

pos_weight = torch.tensor(
    pos_weight_value,
    dtype=torch.float32,
    device=device
)

criterion = nn.BCEWithLogitsLoss(
    pos_weight=pos_weight
)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.001,
    weight_decay=0.0001
)

print("SCER-Mamba ready for training")
```

```python
import numpy as np
import time

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score
)

EPOCHS = 3

best_pr_auc = -1
best_model_path = "/content/best_scer_mamba.pt"

for epoch in range(EPOCHS):

    start_time = time.time()

    # =========================
    # TRAINING
    # =========================
    model.train()

    running_loss = 0.0

    for batch_num, (sequences, static, targets) in enumerate(train_loader):

        sequences = sequences.to(
            device,
            non_blocking=True
        )

        static = static.to(
            device,
            non_blocking=True
        )

        targets = targets.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad()

        logits, attention = model(
            sequences,
            static
        )

        loss = criterion(
            logits,
            targets
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        running_loss += loss.item()

        # Show progress every 50 batches
        if (batch_num + 1) % 50 == 0:

            print(
                f"Epoch {epoch+1} | "
                f"Batch {batch_num+1}/{len(train_loader)} | "
                f"Loss {loss.item():.4f}"
            )

    train_loss = running_loss / len(train_loader)

    # =========================
    # VALIDATION
    # =========================
    model.eval()

    val_probs = []
    val_labels = []

    with torch.no_grad():

        for sequences, static, targets in val_loader:

            sequences = sequences.to(
                device,
                non_blocking=True
            )

            static = static.to(
                device,
                non_blocking=True
            )

            logits, attention = model(
                sequences,
                static
            )

            probabilities = torch.sigmoid(logits)

            val_probs.extend(
                probabilities.cpu().numpy()
            )

            val_labels.extend(
                targets.numpy()
            )

    val_probs = np.asarray(val_probs)
    val_labels = np.asarray(val_labels)

    roc_auc = roc_auc_score(
        val_labels,
        val_probs
    )

    pr_auc = average_precision_score(
        val_labels,
        val_probs
    )

    elapsed = time.time() - start_time

    print("\n--------------------------------")
    print(f"Epoch {epoch+1}/{EPOCHS}")
    print(f"Train Loss : {train_loss:.4f}")
    print(f"ROC-AUC    : {roc_auc:.4f}")
    print(f"PR-AUC     : {pr_auc:.4f}")
    print(f"Time       : {elapsed/60:.2f} minutes")
    print("--------------------------------\n")

    if pr_auc > best_pr_auc:

        best_pr_auc = pr_auc

        torch.save(
            model.state_dict(),
            best_model_path
        )

        print("Best SCER-Mamba model saved.\n")


print("========================================")
print("SCER-MAMBA TRAINING COMPLETED")
print("========================================")
print("Best validation PR-AUC:", best_pr_auc)
```

```python
import torch
import torch.nn as nn
from transformers import MambaConfig, MambaModel

class SCERMambaV2(nn.Module):

    def __init__(self):
        super().__init__()

        hidden_size = 64

        self.static_encoder = nn.Sequential(
            nn.Linear(12, 32),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(32, hidden_size),
            nn.ReLU()
        )

        self.temporal_projection = nn.Linear(7, hidden_size)

        self.gamma = nn.Linear(hidden_size, hidden_size)
        self.beta = nn.Linear(hidden_size, hidden_size)

        config = MambaConfig(
            vocab_size=2,
            hidden_size=64,
            state_size=16,
            num_hidden_layers=2,
            expand=2,
            conv_kernel=4,
            use_cache=False
        )

        self.mamba = MambaModel(config)

        self.temporal_attention = nn.Linear(hidden_size, 1)
        self.risk_event_attention = nn.Linear(3, 1, bias=False)

        self.fusion_gate = nn.Linear(
            hidden_size * 2,
            hidden_size
        )

        self.classifier = nn.Sequential(
            nn.Linear(hidden_size, 16),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(16, 1)
        )

    def forward(self, sequence, static):

        static_embedding = self.static_encoder(static)

        temporal = self.temporal_projection(sequence)

        gamma = torch.tanh(
            self.gamma(static_embedding)
        ).unsqueeze(1)

        beta = self.beta(
            static_embedding
        ).unsqueeze(1)

        conditioned = temporal * (1 + gamma) + beta

        mamba_output = self.mamba(
            inputs_embeds=conditioned,
            use_cache=False
        ).last_hidden_state

        risk_events = sequence[:, :, 4:7]

        temporal_scores = self.temporal_attention(
            mamba_output
        ).squeeze(-1)

        event_scores = self.risk_event_attention(
            risk_events
        ).squeeze(-1)

        attention = torch.softmax(
            temporal_scores + event_scores,
            dim=1
        )

        temporal_embedding = torch.sum(
            mamba_output * attention.unsqueeze(-1),
            dim=1
        )

        fusion_input = torch.cat(
            [temporal_embedding, static_embedding],
            dim=1
        )

        gate = torch.sigmoid(
            self.fusion_gate(fusion_input)
        )

        fused = (
            gate * temporal_embedding
            + (1 - gate) * static_embedding
        )

        logits = self.classifier(fused).squeeze(-1)

        return logits, attention

print("SCER-Mamba V2 created successfully")
```

```python
import torch
from torch.utils.data import (
    TensorDataset,
    DataLoader,
    WeightedRandomSampler
)

# ---------------------------------------------------------
# Training dataset
# ---------------------------------------------------------

train_dataset = TensorDataset(
    torch.tensor(X_seq_train, dtype=torch.float32),
    torch.tensor(X_static_train, dtype=torch.float32),
    torch.tensor(y_train, dtype=torch.float32)
)

val_dataset = TensorDataset(
    torch.tensor(X_seq_val, dtype=torch.float32),
    torch.tensor(X_static_val, dtype=torch.float32),
    torch.tensor(y_val, dtype=torch.float32)
)

test_dataset = TensorDataset(
    torch.tensor(X_seq_test, dtype=torch.float32),
    torch.tensor(X_static_test, dtype=torch.float32),
    torch.tensor(y_test, dtype=torch.float32)
)


# ---------------------------------------------------------
# WeightedRandomSampler
# ---------------------------------------------------------

class_counts = np.bincount(
    y_train.astype(int)
)

print("Class counts:", class_counts)

class_weights = 1.0 / class_counts

sample_weights = class_weights[
    y_train.astype(int)
]

sample_weights = torch.tensor(
    sample_weights,
    dtype=torch.double
)

sampler = WeightedRandomSampler(
    weights=sample_weights,
    num_samples=len(sample_weights),
    replacement=True
)


# ---------------------------------------------------------
# DataLoaders
# ---------------------------------------------------------

train_loader = DataLoader(
    train_dataset,
    batch_size=512,
    sampler=sampler,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=1024,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=1024,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("WeightedRandomSampler created")
print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

from transformers import (
    MambaConfig,
    MambaModel
)


# ============================================================
# FOCAL LOSS
# ============================================================

class FocalLoss(nn.Module):

    def __init__(self, gamma=2.0):
        super().__init__()
        self.gamma = gamma

    def forward(self, logits, targets):

        bce = F.binary_cross_entropy_with_logits(
            logits,
            targets,
            reduction="none"
        )

        pt = torch.exp(-bce)

        focal_loss = (
            (1 - pt) ** self.gamma
        ) * bce

        return focal_loss.mean()


# ============================================================
# IMPROVED SCER-MAMBA
# ============================================================

class ImprovedSCERMamba(nn.Module):

    def __init__(
        self,
        temporal_features=7,
        static_features=12,
        hidden_size=64
    ):

        super().__init__()


        # ----------------------------------------------------
        # Static encoder
        # ----------------------------------------------------

        self.static_encoder = nn.Sequential(
            nn.Linear(static_features, 32),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(32, hidden_size),
            nn.ReLU()
        )


        # ----------------------------------------------------
        # Temporal projection
        # ----------------------------------------------------

        self.temporal_projection = nn.Linear(
            temporal_features,
            hidden_size
        )


        # ----------------------------------------------------
        # Static conditioning
        # ----------------------------------------------------

        self.gamma_layer = nn.Linear(
            hidden_size,
            hidden_size
        )

        self.beta_layer = nn.Linear(
            hidden_size,
            hidden_size
        )


        # ----------------------------------------------------
        # Mamba encoder
        # ----------------------------------------------------

        mamba_config = MambaConfig(
            vocab_size=2,
            hidden_size=64,
            state_size=16,
            num_hidden_layers=2,
            expand=2,
            conv_kernel=4,
            use_cache=False
        )

        self.mamba = MambaModel(
            mamba_config
        )


        # ----------------------------------------------------
        # Risk-event attention
        # ----------------------------------------------------

        self.temporal_attention = nn.Linear(
            hidden_size,
            1
        )

        self.risk_event_attention = nn.Linear(
            3,
            1,
            bias=False
        )


        # ----------------------------------------------------
        # Adaptive gated fusion
        # ----------------------------------------------------

        self.fusion_gate = nn.Linear(
            hidden_size * 2,
            hidden_size
        )


        # ----------------------------------------------------
        # Classifier
        # ----------------------------------------------------

        self.classifier = nn.Sequential(
            nn.Linear(hidden_size, 16),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(16, 1)
        )


    def forward(
        self,
        sequence,
        static
    ):

        # Static representation
        static_embedding = self.static_encoder(
            static
        )

        # Temporal projection
        temporal = self.temporal_projection(
            sequence
        )

        # Static conditioning
        gamma = torch.tanh(
            self.gamma_layer(
                static_embedding
            )
        ).unsqueeze(1)

        beta = self.beta_layer(
            static_embedding
        ).unsqueeze(1)

        conditioned_sequence = (
            temporal * (1 + gamma)
            + beta
        )

        # Mamba
        mamba_output = self.mamba(
            inputs_embeds=conditioned_sequence,
            use_cache=False
        ).last_hidden_state

        # Risk-event features
        risk_events = sequence[:, :, 4:7]

        temporal_scores = self.temporal_attention(
            mamba_output
        ).squeeze(-1)

        event_scores = self.risk_event_attention(
            risk_events
        ).squeeze(-1)

        attention_scores = (
            temporal_scores
            + event_scores
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=1
        )

        temporal_embedding = torch.sum(
            mamba_output
            * attention_weights.unsqueeze(-1),
            dim=1
        )

        # Fusion
        fusion_input = torch.cat(
            [
                temporal_embedding,
                static_embedding
            ],
            dim=1
        )

        gate = torch.sigmoid(
            self.fusion_gate(
                fusion_input
            )
        )

        fused_embedding = (
            gate * temporal_embedding
            +
            (1 - gate) * static_embedding
        )

        # Prediction
        logits = self.classifier(
            fused_embedding
        ).squeeze(-1)

        return logits, attention_weights


print("Focal Loss created")
print("Improved SCER-Mamba created")
```

```python
import random
import numpy as np
import torch

# ============================================================
# REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

print("Device:", device)


# ============================================================
# MODEL
# ============================================================

model = ImprovedSCERMamba().to(device)


# ============================================================
# FOCAL LOSS
# ============================================================

criterion = FocalLoss(
    gamma=2.0
)


# ============================================================
# OPTIMIZER
# ============================================================

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.0003,
    weight_decay=0.0001
)


# ============================================================
# EARLY STOPPING SETTINGS
# ============================================================

MAX_EPOCHS = 20
PATIENCE = 3

best_pr_auc = -1
epochs_without_improvement = 0

BEST_MODEL_PATH = (
    "/content/best_improved_scer_mamba.pt"
)


print("Focal Loss ready")
print("Optimizer ready")
print("Early stopping patience:", PATIENCE)
print("Maximum epochs:", MAX_EPOCHS)
print("Improved SCER-Mamba ready for training")
```

```python
import time
import numpy as np

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score
)

history = []

for epoch in range(MAX_EPOCHS):

    start_time = time.time()

    # ========================================================
    # TRAIN
    # ========================================================

    model.train()

    running_loss = 0.0

    for batch_num, (sequences, static, targets) in enumerate(train_loader):

        sequences = sequences.to(
            device,
            non_blocking=True
        )

        static = static.to(
            device,
            non_blocking=True
        )

        targets = targets.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad()

        logits, attention = model(
            sequences,
            static
        )

        loss = criterion(
            logits,
            targets
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        running_loss += loss.item()

        if (batch_num + 1) % 50 == 0:
            print(
                f"Epoch {epoch+1} | "
                f"Batch {batch_num+1}/{len(train_loader)} | "
                f"Loss: {loss.item():.4f}"
            )

    train_loss = (
        running_loss
        / len(train_loader)
    )


    # ========================================================
    # VALIDATION
    # ========================================================

    model.eval()

    val_probs = []
    val_targets = []

    with torch.no_grad():

        for sequences, static, targets in val_loader:

            sequences = sequences.to(
                device,
                non_blocking=True
            )

            static = static.to(
                device,
                non_blocking=True
            )

            logits, attention = model(
                sequences,
                static
            )

            probs = torch.sigmoid(
                logits
            )

            val_probs.extend(
                probs.cpu().numpy()
            )

            val_targets.extend(
                targets.numpy()
            )

    val_probs = np.asarray(
        val_probs
    )

    val_targets = np.asarray(
        val_targets
    )

    roc_auc = roc_auc_score(
        val_targets,
        val_probs
    )

    pr_auc = average_precision_score(
        val_targets,
        val_probs
    )

    elapsed = (
        time.time()
        - start_time
    )


    print("\n------------------------------------")
    print(f"Epoch {epoch+1}/{MAX_EPOCHS}")
    print(f"Train Loss : {train_loss:.4f}")
    print(f"ROC-AUC    : {roc_auc:.4f}")
    print(f"PR-AUC     : {pr_auc:.4f}")
    print(f"Time       : {elapsed/60:.2f} minutes")
    print("------------------------------------")


    history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "roc_auc": roc_auc,
        "pr_auc": pr_auc
    })


    # ========================================================
    # EARLY STOPPING
    # ========================================================

    if pr_auc > best_pr_auc:

        best_pr_auc = pr_auc

        epochs_without_improvement = 0

        torch.save(
            model.state_dict(),
            BEST_MODEL_PATH
        )

        print("Best model saved.\n")

    else:

        epochs_without_improvement += 1

        print(
            "No PR-AUC improvement:",
            epochs_without_improvement,
            "/",
            PATIENCE
        )

        if epochs_without_improvement >= PATIENCE:

            print("\nEARLY STOPPING TRIGGERED")
            break


print("\n========================================")
print("IMPROVED SCER-MAMBA TRAINING COMPLETE")
print("========================================")

print(
    "Best validation PR-AUC:",
    best_pr_auc
)
```

```python
import torch
import torch.nn as nn

from transformers import MambaConfig, MambaModel


class SCERMambaNoGate(nn.Module):

    def __init__(
        self,
        temporal_features=7,
        static_features=12,
        hidden_size=64
    ):
        super().__init__()

        # ====================================================
        # STATIC ENCODER
        # ====================================================

        self.static_encoder = nn.Sequential(
            nn.Linear(static_features, 32),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(32, hidden_size),
            nn.ReLU()
        )

        # ====================================================
        # TEMPORAL PROJECTION
        # ====================================================

        self.temporal_projection = nn.Linear(
            temporal_features,
            hidden_size
        )

        # ====================================================
        # STATIC CONDITIONING
        # ====================================================

        self.gamma_layer = nn.Linear(
            hidden_size,
            hidden_size
        )

        self.beta_layer = nn.Linear(
            hidden_size,
            hidden_size
        )

        # ====================================================
        # MAMBA
        # ====================================================

        mamba_config = MambaConfig(
            vocab_size=2,
            hidden_size=64,
            state_size=16,
            num_hidden_layers=2,
            expand=2,
            conv_kernel=4,
            use_cache=False
        )

        self.mamba = MambaModel(
            mamba_config
        )

        # ====================================================
        # RISK-EVENT ATTENTION
        # ====================================================

        self.temporal_attention = nn.Linear(
            hidden_size,
            1
        )

        self.risk_event_attention = nn.Linear(
            3,
            1,
            bias=False
        )

        # ====================================================
        # NO ADAPTIVE GATE
        # Direct temporal + static concatenation
        # ====================================================

        self.classifier = nn.Sequential(
            nn.Linear(hidden_size * 2, 16),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(16, 1)
        )


    def forward(
        self,
        sequence,
        static
    ):

        # Static representation
        static_embedding = self.static_encoder(
            static
        )

        # Temporal projection
        temporal = self.temporal_projection(
            sequence
        )

        # Static conditioning
        gamma = torch.tanh(
            self.gamma_layer(
                static_embedding
            )
        ).unsqueeze(1)

        beta = self.beta_layer(
            static_embedding
        ).unsqueeze(1)

        conditioned_sequence = (
            temporal * (1 + gamma)
            + beta
        )

        # Mamba
        mamba_output = self.mamba(
            inputs_embeds=conditioned_sequence,
            use_cache=False
        ).last_hidden_state

        # Risk-event attention
        risk_events = sequence[:, :, 4:7]

        temporal_scores = self.temporal_attention(
            mamba_output
        ).squeeze(-1)

        event_scores = self.risk_event_attention(
            risk_events
        ).squeeze(-1)

        attention_scores = (
            temporal_scores
            + event_scores
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=1
        )

        temporal_embedding = torch.sum(
            mamba_output
            * attention_weights.unsqueeze(-1),
            dim=1
        )

        # ====================================================
        # DIRECT FUSION - NO GATE
        # ====================================================

        fused_embedding = torch.cat(
            [
                temporal_embedding,
                static_embedding
            ],
            dim=1
        )

        logits = self.classifier(
            fused_embedding
        ).squeeze(-1)

        return logits, attention_weights


print("======================================")
print("EXPERIMENT C MODEL CREATED")
print("Mamba                  : YES")
print("Static Conditioning    : YES")
print("Risk-Event Attention   : YES")
print("Adaptive Fusion Gate   : NO")
print("======================================")
```

```python
import torch
import torch.nn as nn

from transformers import MambaConfig, MambaModel


class SCERMambaNoConditioning(nn.Module):

    def __init__(
        self,
        temporal_features=7,
        static_features=12,
        hidden_size=64
    ):
        super().__init__()

        # ====================================================
        # STATIC ENCODER
        # ====================================================

        self.static_encoder = nn.Sequential(
            nn.Linear(static_features, 32),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(32, hidden_size),
            nn.ReLU()
        )

        # ====================================================
        # TEMPORAL PROJECTION
        # ====================================================

        self.temporal_projection = nn.Linear(
            temporal_features,
            hidden_size
        )

        # ====================================================
        # MAMBA
        # ====================================================

        mamba_config = MambaConfig(
            vocab_size=2,
            hidden_size=64,
            state_size=16,
            num_hidden_layers=2,
            expand=2,
            conv_kernel=4,
            use_cache=False
        )

        self.mamba = MambaModel(
            mamba_config
        )

        # ====================================================
        # RISK-EVENT ATTENTION
        # ====================================================

        self.temporal_attention = nn.Linear(
            hidden_size,
            1
        )

        self.risk_event_attention = nn.Linear(
            3,
            1,
            bias=False
        )

        # ====================================================
        # ADAPTIVE GATED FUSION
        # ====================================================

        self.fusion_gate = nn.Linear(
            hidden_size * 2,
            hidden_size
        )

        # ====================================================
        # CLASSIFIER
        # ====================================================

        self.classifier = nn.Sequential(
            nn.Linear(hidden_size, 16),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(16, 1)
        )


    def forward(
        self,
        sequence,
        static
    ):

        # Static representation
        static_embedding = self.static_encoder(
            static
        )

        # Temporal projection
        temporal = self.temporal_projection(
            sequence
        )

        # ====================================================
        # NO STATIC CONDITIONING
        # Temporal input goes directly into Mamba
        # ====================================================

        mamba_output = self.mamba(
            inputs_embeds=temporal,
            use_cache=False
        ).last_hidden_state

        # ====================================================
        # RISK-EVENT ATTENTION
        # ====================================================

        risk_events = sequence[:, :, 4:7]

        temporal_scores = self.temporal_attention(
            mamba_output
        ).squeeze(-1)

        event_scores = self.risk_event_attention(
            risk_events
        ).squeeze(-1)

        attention_scores = (
            temporal_scores
            + event_scores
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=1
        )

        temporal_embedding = torch.sum(
            mamba_output
            * attention_weights.unsqueeze(-1),
            dim=1
        )

        # ====================================================
        # ADAPTIVE GATED FUSION
        # ====================================================

        fusion_input = torch.cat(
            [
                temporal_embedding,
                static_embedding
            ],
            dim=1
        )

        gate = torch.sigmoid(
            self.fusion_gate(
                fusion_input
            )
        )

        fused_embedding = (
            gate * temporal_embedding
            +
            (1 - gate) * static_embedding
        )

        # Prediction
        logits = self.classifier(
            fused_embedding
        ).squeeze(-1)

        return logits, attention_weights


print("======================================")
print("EXPERIMENT D MODEL CREATED")
print("Mamba                  : YES")
print("Static Conditioning    : NO")
print("Risk-Event Attention   : YES")
print("Adaptive Fusion Gate   : YES")
print("======================================")
```

```python
import random
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import (
    TensorDataset,
    DataLoader
)

from transformers import (
    MambaConfig,
    MambaModel
)

# ============================================================
# REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

print("Device:", device)


# ============================================================
# DATASETS
# ============================================================

train_dataset_F = TensorDataset(
    torch.tensor(X_seq_train, dtype=torch.float32),
    torch.tensor(X_static_train, dtype=torch.float32),
    torch.tensor(y_train, dtype=torch.float32)
)

val_dataset_F = TensorDataset(
    torch.tensor(X_seq_val, dtype=torch.float32),
    torch.tensor(X_static_val, dtype=torch.float32),
    torch.tensor(y_val, dtype=torch.float32)
)

test_dataset_F = TensorDataset(
    torch.tensor(X_seq_test, dtype=torch.float32),
    torch.tensor(X_static_test, dtype=torch.float32),
    torch.tensor(y_test, dtype=torch.float32)
)


train_loader_F = DataLoader(
    train_dataset_F,
    batch_size=512,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader_F = DataLoader(
    val_dataset_F,
    batch_size=1024,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader_F = DataLoader(
    test_dataset_F,
    batch_size=1024,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)


# ============================================================
# FOCAL LOSS
# ============================================================

class FocalLoss(nn.Module):

    def __init__(self, gamma=2.0):
        super().__init__()
        self.gamma = gamma

    def forward(self, logits, targets):

        bce = F.binary_cross_entropy_with_logits(
            logits,
            targets,
            reduction="none"
        )

        pt = torch.exp(-bce)

        return (
            ((1 - pt) ** self.gamma) * bce
        ).mean()


# ============================================================
# ENHANCED SCER-MAMBA
# Experiment D architecture
# No Static Conditioning
# ============================================================

class EnhancedSCERMamba(nn.Module):

    def __init__(
        self,
        temporal_features=13,
        static_features=12,
        hidden_size=64,
        num_layers=2,
        dropout=0.20
    ):
        super().__init__()

        # Static encoder
        self.static_encoder = nn.Sequential(
            nn.Linear(static_features, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, hidden_size),
            nn.ReLU()
        )

        # 13 temporal features -> Mamba hidden dimension
        self.temporal_projection = nn.Linear(
            temporal_features,
            hidden_size
        )

        # Mamba
        config = MambaConfig(
            vocab_size=2,
            hidden_size=hidden_size,
            state_size=16,
            num_hidden_layers=num_layers,
            expand=2,
            conv_kernel=4,
            use_cache=False
        )

        self.mamba = MambaModel(config)

        # Temporal attention
        self.temporal_attention = nn.Linear(
            hidden_size,
            1
        )

        # Existing delinquency event indicators:
        # columns 4, 5, 6
        self.risk_event_attention = nn.Linear(
            3,
            1,
            bias=False
        )

        # Adaptive gated fusion
        self.fusion_gate = nn.Linear(
            hidden_size * 2,
            hidden_size
        )

        # Classifier
        self.classifier = nn.Sequential(
            nn.Linear(hidden_size, 16),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(16, 1)
        )


    def forward(self, sequence, static):

        static_embedding = self.static_encoder(
            static
        )

        temporal = self.temporal_projection(
            sequence
        )

        # NO static conditioning
        mamba_output = self.mamba(
            inputs_embeds=temporal,
            use_cache=False
        ).last_hidden_state

        # Original delinquency event indicators
        risk_events = sequence[:, :, 4:7]

        temporal_scores = self.temporal_attention(
            mamba_output
        ).squeeze(-1)

        event_scores = self.risk_event_attention(
            risk_events
        ).squeeze(-1)

        attention_scores = (
            temporal_scores
            + event_scores
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=1
        )

        temporal_embedding = torch.sum(
            mamba_output
            * attention_weights.unsqueeze(-1),
            dim=1
        )

        fusion_input = torch.cat(
            [
                temporal_embedding,
                static_embedding
            ],
            dim=1
        )

        gate = torch.sigmoid(
            self.fusion_gate(
                fusion_input
            )
        )

        fused_embedding = (
            gate * temporal_embedding
            +
            (1 - gate) * static_embedding
        )

        logits = self.classifier(
            fused_embedding
        ).squeeze(-1)

        return logits, attention_weights


# ============================================================
# MODEL
# ============================================================

model_F = EnhancedSCERMamba(
    temporal_features=13,
    hidden_size=64,
    num_layers=2,
    dropout=0.20
).to(device)

criterion_F = FocalLoss(
    gamma=2.0
)

optimizer_F = torch.optim.AdamW(
    model_F.parameters(),
    lr=0.0003,
    weight_decay=0.0001
)

print("\n======================================")
print("EXPERIMENT F READY")
print("Temporal features    : 13")
print("Static features      : 12")
print("Mamba hidden size    : 64")
print("Mamba layers         : 2")
print("Static Conditioning  : NO")
print("Risk-Event Attention : YES")
print("Adaptive Gate        : YES")
print("Focal Loss           : YES")
print("======================================")
```

```python
import torch
import torch.nn as nn
from transformers import MambaConfig, MambaModel


class EnhancedRiskAttentionSCERMamba(nn.Module):

    def __init__(
        self,
        temporal_features=13,
        static_features=12,
        hidden_size=64,
        num_layers=2,
        dropout=0.20
    ):
        super().__init__()

        # Static encoder
        self.static_encoder = nn.Sequential(
            nn.Linear(static_features, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, hidden_size),
            nn.ReLU()
        )

        # Temporal projection
        self.temporal_projection = nn.Linear(
            temporal_features,
            hidden_size
        )

        # Mamba encoder
        config = MambaConfig(
            vocab_size=2,
            hidden_size=hidden_size,
            state_size=16,
            num_hidden_layers=num_layers,
            expand=2,
            conv_kernel=4,
            use_cache=False
        )

        self.mamba = MambaModel(config)

        # Normal temporal attention
        self.temporal_attention = nn.Linear(
            hidden_size,
            1
        )

        # ====================================================
        # ENHANCED RISK-EVENT ATTENTION
        # 9 risk-related temporal inputs
        # ====================================================

        self.risk_event_attention = nn.Sequential(
            nn.Linear(9, 16),
            nn.ReLU(),
            nn.Linear(16, 1)
        )

        # Adaptive gated fusion
        self.fusion_gate = nn.Linear(
            hidden_size * 2,
            hidden_size
        )

        # Classifier
        self.classifier = nn.Sequential(
            nn.Linear(hidden_size, 16),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(16, 1)
        )


    def forward(self, sequence, static):

        static_embedding = self.static_encoder(
            static
        )

        temporal = self.temporal_projection(
            sequence
        )

        mamba_output = self.mamba(
            inputs_embeds=temporal,
            use_cache=False
        ).last_hidden_state

        # ====================================================
        # Risk features:
        # 4  is_30plus_delinquent
        # 5  is_60plus_delinquent
        # 6  is_90plus_delinquent
        # 7  delinquency_change
        # 8  delinquency_worsening
        # 9  cumulative_max_delinquency
        # 10 consecutive_30plus_months
        # 11 recent_3m_30plus_count
        # 12 monthly_upb_change
        # ====================================================

        risk_events = sequence[:, :, 4:13]

        temporal_scores = self.temporal_attention(
            mamba_output
        ).squeeze(-1)

        event_scores = self.risk_event_attention(
            risk_events
        ).squeeze(-1)

        attention_scores = (
            temporal_scores
            + event_scores
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=1
        )

        temporal_embedding = torch.sum(
            mamba_output
            * attention_weights.unsqueeze(-1),
            dim=1
        )

        fusion_input = torch.cat(
            [
                temporal_embedding,
                static_embedding
            ],
            dim=1
        )

        gate = torch.sigmoid(
            self.fusion_gate(
                fusion_input
            )
        )

        fused_embedding = (
            gate * temporal_embedding
            +
            (1 - gate) * static_embedding
        )

        logits = self.classifier(
            fused_embedding
        ).squeeze(-1)

        return logits, attention_weights


print("======================================")
print("EXPERIMENT H MODEL READY")
print("Temporal features       : 13")
print("Risk-attention features : 9")
print("Hidden size             : 64")
print("Mamba layers            : 2")
print("Static conditioning     : NO")
print("Enhanced risk attention : YES")
print("Adaptive gate           : YES")
print("======================================")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class AlphaFocalLoss(nn.Module):
    def __init__(self, alpha=0.75, gamma=2.0):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma

    def forward(self, logits, targets):

        targets = targets.float()

        bce = F.binary_cross_entropy_with_logits(
            logits,
            targets,
            reduction="none"
        )

        probs = torch.sigmoid(logits)

        pt = (
            probs * targets
            + (1 - probs) * (1 - targets)
        )

        alpha_t = (
            self.alpha * targets
            + (1 - self.alpha) * (1 - targets)
        )

        focal_weight = (
            alpha_t *
            (1 - pt) ** self.gamma
        )

        return (focal_weight * bce).mean()


print("✅ Alpha-weighted Focal Loss ready")
```

```python
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import DataLoader
from transformers import MambaConfig, MambaModel
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    precision_recall_curve
)

# ============================================================
# G1 ARCHITECTURE
# 13 temporal features
# Risk attention uses original cols 4:7 only
# ============================================================

class G1EnhancedSCERMamba(nn.Module):

    def __init__(
        self,
        temporal_features=13,
        static_features=12,
        hidden_size=64,
        num_layers=2,
        dropout=0.20
    ):
        super().__init__()

        self.static_encoder = nn.Sequential(
            nn.Linear(static_features, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, hidden_size),
            nn.ReLU()
        )

        self.temporal_projection = nn.Linear(
            temporal_features,
            hidden_size
        )

        config = MambaConfig(
            vocab_size=2,
            hidden_size=hidden_size,
            state_size=16,
            num_hidden_layers=num_layers,
            expand=2,
            conv_kernel=4,
            use_cache=False
        )

        self.mamba = MambaModel(config)

        self.temporal_attention = nn.Linear(
            hidden_size,
            1
        )

        # Original 3-feature risk attention
        self.risk_event_attention = nn.Sequential(
            nn.Linear(3, 16),
            nn.ReLU(),
            nn.Linear(16, 1)
        )

        self.fusion_gate = nn.Linear(
            hidden_size * 2,
            hidden_size
        )

        self.classifier = nn.Sequential(
            nn.Linear(hidden_size, 16),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(16, 1)
        )

    def forward(self, sequence, static):

        static_embedding = self.static_encoder(static)

        temporal = self.temporal_projection(sequence)

        mamba_output = self.mamba(
            inputs_embeds=temporal,
            use_cache=False
        ).last_hidden_state

        temporal_scores = self.temporal_attention(
            mamba_output
        ).squeeze(-1)

        # Only 30+, 60+, 90+ features
        risk_events = sequence[:, :, 4:7]

        event_scores = self.risk_event_attention(
            risk_events
        ).squeeze(-1)

        attention_weights = torch.softmax(
            temporal_scores + event_scores,
            dim=1
        )

        temporal_embedding = torch.sum(
            mamba_output *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        fusion_input = torch.cat(
            [temporal_embedding, static_embedding],
            dim=1
        )

        gate = torch.sigmoid(
            self.fusion_gate(fusion_input)
        )

        fused = (
            gate * temporal_embedding
            +
            (1 - gate) * static_embedding
        )

        logits = self.classifier(
            fused
        ).squeeze(-1)

        return logits, attention_weights


# ============================================================
# TRAIN ONE ALPHA VALUE
# ============================================================

def train_alpha(alpha, seed=42):

    print("\n========================================")
    print("ALPHA:", alpha)
    print("========================================")

    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    # Fresh loader so every alpha gets same shuffle seed
    generator = torch.Generator()
    generator.manual_seed(seed)

    train_loader = DataLoader(
        train_dataset_F,
        batch_size=256,
        shuffle=True,
        generator=generator,
        num_workers=2,
        pin_memory=True
    )

    model = G1EnhancedSCERMamba(
        temporal_features=13,
        static_features=12,
        hidden_size=64,
        num_layers=2,
        dropout=0.20
    ).to(device)

    criterion = AlphaFocalLoss(
        alpha=alpha,
        gamma=2.0
    )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=0.0003,
        weight_decay=0.0001
    )

    best_pr = -1
    best_roc = -1
    best_f1 = -1
    best_threshold = None

    patience = 3
    no_improvement = 0

    model_path = f"/content/experiment_I_alpha_{alpha}.pt"

    for epoch in range(15):

        model.train()

        for sequence, static, targets in train_loader:

            sequence = sequence.to(device)
            static = static.to(device)
            targets = targets.to(device)

            optimizer.zero_grad()

            logits, _ = model(
                sequence,
                static
            )

            loss = criterion(
                logits,
                targets
            )

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                1.0
            )

            optimizer.step()

        # ==========================
        # VALIDATION
        # ==========================

        model.eval()

        probs_all = []
        labels_all = []

        with torch.no_grad():

            for sequence, static, targets in val_loader_F:

                sequence = sequence.to(device)
                static = static.to(device)

                logits, _ = model(
                    sequence,
                    static
                )

                probs = torch.sigmoid(logits)

                probs_all.extend(
                    probs.cpu().numpy()
                )

                labels_all.extend(
                    targets.numpy()
                )

        probs_all = np.asarray(probs_all)
        labels_all = np.asarray(labels_all)

        pr_auc = average_precision_score(
            labels_all,
            probs_all
        )

        roc_auc = roc_auc_score(
            labels_all,
            probs_all
        )

        precision, recall, thresholds = (
            precision_recall_curve(
                labels_all,
                probs_all
            )
        )

        f1s = (
            2 * precision[:-1] * recall[:-1]
            /
            (
                precision[:-1]
                + recall[:-1]
                + 1e-10
            )
        )

        idx = np.argmax(f1s)

        epoch_f1 = float(f1s[idx])
        threshold = float(thresholds[idx])

        print(
            f"Epoch {epoch+1:02d} | "
            f"PR-AUC {pr_auc:.4f} | "
            f"ROC {roc_auc:.4f} | "
            f"F1 {epoch_f1:.4f}"
        )

        if pr_auc > best_pr:

            best_pr = pr_auc
            best_roc = roc_auc
            best_f1 = epoch_f1
            best_threshold = threshold

            no_improvement = 0

            torch.save(
                model.state_dict(),
                model_path
            )

        else:

            no_improvement += 1

            if no_improvement >= patience:
                print("Early stopping")
                break

    return {
        "alpha": alpha,
        "val_pr_auc": best_pr,
        "val_roc_auc": best_roc,
        "val_f1": best_f1,
        "threshold": best_threshold,
        "model_path": model_path
    }


# ============================================================
# EXPERIMENT I SEARCH
# ============================================================

alpha_values = [
    0.60,
    0.75,
    0.90,
    0.95
]

results_I = []

for alpha in alpha_values:

    result = train_alpha(alpha)

    results_I.append(result)


results_I_df = pd.DataFrame(
    results_I
).sort_values(
    "val_pr_auc",
    ascending=False
)

print("\n========================================")
print("EXPERIMENT I RESULTS")
print("========================================")

print(results_I_df)

print("\n========================================")
print("BEST ALPHA")
print("========================================")

print(results_I_df.iloc[0])

results_I_df.to_csv(
    "/content/experiment_I_alpha_tuning.csv",
    index=False
)

print("\nSaved:")
print("/content/experiment_I_alpha_tuning.csv")
```

```python
import random
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F

from transformers import MambaConfig, MambaModel
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    precision_recall_curve
)

# ============================================================
# DEVICE + SEED
# ============================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

print("Using device:", device)


# ============================================================
# FOCAL LOSS
# ============================================================

class FocalLoss(nn.Module):

    def __init__(self, gamma=2.0):
        super().__init__()
        self.gamma = gamma

    def forward(self, logits, targets):

        targets = targets.float()

        bce = F.binary_cross_entropy_with_logits(
            logits,
            targets,
            reduction="none"
        )

        probs = torch.sigmoid(logits)

        pt = (
            probs * targets
            +
            (1 - probs) * (1 - targets)
        )

        return (
            ((1 - pt) ** self.gamma) * bce
        ).mean()


# ============================================================
# G1 ENHANCED SCER-MAMBA
# ============================================================

class G1EnhancedSCERMamba(nn.Module):

    def __init__(
        self,
        temporal_features=13,
        static_features=12,
        hidden_size=64,
        num_layers=2,
        dropout=0.20
    ):
        super().__init__()

        self.static_encoder = nn.Sequential(
            nn.Linear(static_features, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, hidden_size),
            nn.ReLU()
        )

        self.temporal_projection = nn.Linear(
            temporal_features,
            hidden_size
        )

        config = MambaConfig(
            vocab_size=2,
            hidden_size=hidden_size,
            state_size=16,
            num_hidden_layers=num_layers,
            expand=2,
            conv_kernel=4,
            use_cache=False
        )

        self.mamba = MambaModel(config)

        self.temporal_attention = nn.Linear(
            hidden_size,
            1
        )

        self.risk_event_attention = nn.Sequential(
            nn.Linear(3, 16),
            nn.ReLU(),
            nn.Linear(16, 1)
        )

        self.fusion_gate = nn.Linear(
            hidden_size * 2,
            hidden_size
        )

        self.classifier = nn.Sequential(
            nn.Linear(hidden_size, 16),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(16, 1)
        )

    def forward(self, sequence, static):

        static_embedding = self.static_encoder(
            static
        )

        temporal = self.temporal_projection(
            sequence
        )

        mamba_output = self.mamba(
            inputs_embeds=temporal,
            use_cache=False
        ).last_hidden_state

        temporal_scores = self.temporal_attention(
            mamba_output
        ).squeeze(-1)

        risk_events = sequence[:, :, 4:7]

        event_scores = self.risk_event_attention(
            risk_events
        ).squeeze(-1)

        attention_weights = torch.softmax(
            temporal_scores + event_scores,
            dim=1
        )

        temporal_embedding = torch.sum(
            mamba_output *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        fusion_input = torch.cat(
            [
                temporal_embedding,
                static_embedding
            ],
            dim=1
        )

        gate = torch.sigmoid(
            self.fusion_gate(
                fusion_input
            )
        )

        fused = (
            gate * temporal_embedding
            +
            (1 - gate) * static_embedding
        )

        logits = self.classifier(
            fused
        ).squeeze(-1)

        return logits, attention_weights


# ============================================================
# J18 MODEL
# ============================================================

model_J18 = G1EnhancedSCERMamba(
    temporal_features=13,
    static_features=12,
    hidden_size=64,
    num_layers=2,
    dropout=0.20
).to(device)

criterion_J18 = FocalLoss(
    gamma=2.0
)

optimizer_J18 = torch.optim.AdamW(
    model_J18.parameters(),
    lr=0.0003,
    weight_decay=0.0001
)

MAX_EPOCHS = 20
PATIENCE = 4

best_pr = -1
best_roc = -1
best_f1 = -1
best_threshold = None

no_improvement = 0

BEST_PATH_J18 = (
    "/content/best_SCER_Mamba_J18.pt"
)

# ============================================================
# TRAIN
# ============================================================

for epoch in range(MAX_EPOCHS):

    model_J18.train()

    total_loss = 0.0

    for sequence, static, targets in train_loader_J18:

        sequence = sequence.to(device)
        static = static.to(device)
        targets = targets.to(device)

        optimizer_J18.zero_grad()

        logits, _ = model_J18(
            sequence,
            static
        )

        loss = criterion_J18(
            logits,
            targets
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model_J18.parameters(),
            1.0
        )

        optimizer_J18.step()

        total_loss += loss.item()

    # ========================================================
    # VALIDATION
    # ========================================================

    model_J18.eval()

    probs_all = []
    labels_all = []

    with torch.no_grad():

        for sequence, static, targets in val_loader_J18:

            sequence = sequence.to(device)
            static = static.to(device)

            logits, _ = model_J18(
                sequence,
                static
            )

            probs = torch.sigmoid(logits)

            probs_all.extend(
                probs.cpu().numpy()
            )

            labels_all.extend(
                targets.numpy()
            )

    probs_all = np.asarray(probs_all)
    labels_all = np.asarray(labels_all)

    pr_auc = average_precision_score(
        labels_all,
        probs_all
    )

    roc_auc = roc_auc_score(
        labels_all,
        probs_all
    )

    precision, recall, thresholds = (
        precision_recall_curve(
            labels_all,
            probs_all
        )
    )

    f1_values = (
        2 * precision[:-1] * recall[:-1]
        /
        (
            precision[:-1]
            + recall[:-1]
            + 1e-10
        )
    )

    idx = np.argmax(f1_values)

    epoch_f1 = float(
        f1_values[idx]
    )

    epoch_threshold = float(
        thresholds[idx]
    )

    print(
        f"Epoch {epoch+1:02d} | "
        f"Loss {total_loss / len(train_loader_J18):.4f} | "
        f"PR-AUC {pr_auc:.4f} | "
        f"ROC {roc_auc:.4f} | "
        f"F1 {epoch_f1:.4f} | "
        f"Threshold {epoch_threshold:.4f}"
    )

    # ========================================================
    # EARLY STOPPING BY PR-AUC
    # ========================================================

    if pr_auc > best_pr:

        best_pr = pr_auc
        best_roc = roc_auc
        best_f1 = epoch_f1
        best_threshold = epoch_threshold

        no_improvement = 0

        torch.save(
            model_J18.state_dict(),
            BEST_PATH_J18
        )

        print("✅ Best J18 model saved")

    else:

        no_improvement += 1

        print(
            f"No improvement: "
            f"{no_improvement}/{PATIENCE}"
        )

        if no_improvement >= PATIENCE:
            print("EARLY STOPPING")
            break


print("\n======================================")
print("J18 TRAINING COMPLETE")
print("======================================")

print("Best validation PR-AUC :", best_pr)
print("ROC-AUC                :", best_roc)
print("F1                     :", best_f1)
print("Threshold              :", best_threshold)
print("Model                   :", BEST_PATH_J18)
```

```python
import random
import numpy as np
import torch
import torch.nn as nn

from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    precision_recall_curve
)

# ============================================================
# STANDARD LSTM BASELINE
# Same temporal + static inputs
# ============================================================

class LSTMBaseline(nn.Module):

    def __init__(
        self,
        temporal_features=13,
        static_features=12,
        hidden_size=64,
        num_layers=2,
        dropout=0.20
    ):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=temporal_features,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True,
            dropout=dropout
        )

        self.static_encoder = nn.Sequential(
            nn.Linear(static_features, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, hidden_size),
            nn.ReLU()
        )

        self.classifier = nn.Sequential(
            nn.Linear(hidden_size * 2, 64),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(64, 1)
        )

    def forward(self, sequence, static):

        lstm_output, (hidden, cell) = self.lstm(
            sequence
        )

        temporal_embedding = hidden[-1]

        static_embedding = self.static_encoder(
            static
        )

        fused = torch.cat(
            [
                temporal_embedding,
                static_embedding
            ],
            dim=1
        )

        logits = self.classifier(
            fused
        ).squeeze(-1)

        return logits


# ============================================================
# J18 TRAINING
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

model_LSTM_J18 = LSTMBaseline(
    temporal_features=13,
    static_features=12,
    hidden_size=64,
    num_layers=2,
    dropout=0.20
).to(device)

criterion = FocalLoss(
    gamma=2.0
)

optimizer = torch.optim.AdamW(
    model_LSTM_J18.parameters(),
    lr=0.0003,
    weight_decay=0.0001
)

MAX_EPOCHS = 20
PATIENCE = 4

best_pr = -1
best_roc = -1
best_f1 = -1
best_threshold = None

no_improvement = 0

BEST_PATH = "/content/best_LSTM_J18.pt"

for epoch in range(MAX_EPOCHS):

    model_LSTM_J18.train()

    total_loss = 0.0

    for sequence, static, targets in train_loader_J18:

        sequence = sequence.to(device)
        static = static.to(device)
        targets = targets.to(device)

        optimizer.zero_grad()

        logits = model_LSTM_J18(
            sequence,
            static
        )

        loss = criterion(
            logits,
            targets
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model_LSTM_J18.parameters(),
            1.0
        )

        optimizer.step()

        total_loss += loss.item()

    # ========================================================
    # VALIDATION
    # ========================================================

    model_LSTM_J18.eval()

    probs_all = []
    labels_all = []

    with torch.no_grad():

        for sequence, static, targets in val_loader_J18:

            sequence = sequence.to(device)
            static = static.to(device)

            logits = model_LSTM_J18(
                sequence,
                static
            )

            probs = torch.sigmoid(logits)

            probs_all.extend(
                probs.cpu().numpy()
            )

            labels_all.extend(
                targets.numpy()
            )

    probs_all = np.asarray(probs_all)
    labels_all = np.asarray(labels_all)

    pr_auc = average_precision_score(
        labels_all,
        probs_all
    )

    roc_auc = roc_auc_score(
        labels_all,
        probs_all
    )

    precision, recall, thresholds = (
        precision_recall_curve(
            labels_all,
            probs_all
        )
    )

    f1_values = (
        2 * precision[:-1] * recall[:-1]
        /
        (
            precision[:-1]
            + recall[:-1]
            + 1e-10
        )
    )

    idx = np.argmax(f1_values)

    epoch_f1 = float(
        f1_values[idx]
    )

    epoch_threshold = float(
        thresholds[idx]
    )

    print(
        f"Epoch {epoch+1:02d} | "
        f"Loss {total_loss / len(train_loader_J18):.4f} | "
        f"PR-AUC {pr_auc:.4f} | "
        f"ROC {roc_auc:.4f} | "
        f"F1 {epoch_f1:.4f}"
    )

    if pr_auc > best_pr:

        best_pr = pr_auc
        best_roc = roc_auc
        best_f1 = epoch_f1
        best_threshold = epoch_threshold

        no_improvement = 0

        torch.save(
            model_LSTM_J18.state_dict(),
            BEST_PATH
        )

        print("✅ Best J18 LSTM saved")

    else:

        no_improvement += 1

        print(
            f"No improvement: "
            f"{no_improvement}/{PATIENCE}"
        )

        if no_improvement >= PATIENCE:
            print("EARLY STOPPING")
            break


print("\n======================================")
print("J18 LSTM VALIDATION COMPLETE")
print("======================================")

print("Best validation PR-AUC :", best_pr)
print("ROC-AUC                :", best_roc)
print("F1                     :", best_f1)
print("Threshold              :", best_threshold)
print("Model                  :", BEST_PATH)
```

```python
import random
import numpy as np
import torch
import torch.nn as nn

from transformers import MambaConfig, MambaModel
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    precision_recall_curve
)

# ============================================================
# EXPERIMENT K MODEL
# ============================================================

class MultiPoolSCERMamba(nn.Module):

    def __init__(
        self,
        temporal_features=13,
        static_features=12,
        hidden_size=64,
        num_layers=2,
        dropout=0.20
    ):
        super().__init__()

        # Static pathway
        self.static_encoder = nn.Sequential(
            nn.Linear(static_features, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, hidden_size),
            nn.ReLU()
        )

        # Temporal projection
        self.temporal_projection = nn.Linear(
            temporal_features,
            hidden_size
        )

        # Mamba
        config = MambaConfig(
            vocab_size=2,
            hidden_size=hidden_size,
            state_size=16,
            num_hidden_layers=num_layers,
            expand=2,
            conv_kernel=4,
            use_cache=False
        )

        self.mamba = MambaModel(config)

        self.output_norm = nn.LayerNorm(
            hidden_size
        )

        # Standard temporal attention
        self.temporal_attention = nn.Linear(
            hidden_size,
            1
        )

        # Risk-event attention
        self.risk_event_attention = nn.Sequential(
            nn.Linear(3, 16),
            nn.ReLU(),
            nn.Linear(16, 1)
        )

        # 4 representations:
        # attention pool
        # mean pool
        # final state
        # static embedding
        self.classifier = nn.Sequential(
            nn.Linear(hidden_size * 4, 128),
            nn.ReLU(),
            nn.Dropout(dropout),

            nn.Linear(128, 32),
            nn.ReLU(),
            nn.Dropout(dropout),

            nn.Linear(32, 1)
        )

    def forward(self, sequence, static):

        static_embedding = self.static_encoder(
            static
        )

        temporal = self.temporal_projection(
            sequence
        )

        mamba_output = self.mamba(
            inputs_embeds=temporal,
            use_cache=False
        ).last_hidden_state

        mamba_output = self.output_norm(
            mamba_output
        )

        # --------------------------------------------
        # Risk-aware attention pooling
        # --------------------------------------------

        temporal_scores = self.temporal_attention(
            mamba_output
        ).squeeze(-1)

        risk_events = sequence[:, :, 4:7]

        risk_scores = self.risk_event_attention(
            risk_events
        ).squeeze(-1)

        attention_weights = torch.softmax(
            temporal_scores + risk_scores,
            dim=1
        )

        attention_pool = torch.sum(
            mamba_output *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        # --------------------------------------------
        # Additional temporal summaries
        # --------------------------------------------

        mean_pool = mamba_output.mean(
            dim=1
        )

        final_state = mamba_output[:, -1, :]

        # --------------------------------------------
        # Direct concatenation instead of gate
        # --------------------------------------------

        combined = torch.cat(
            [
                attention_pool,
                mean_pool,
                final_state,
                static_embedding
            ],
            dim=1
        )

        logits = self.classifier(
            combined
        ).squeeze(-1)

        return logits, attention_weights


# ============================================================
# TRAINING SETUP
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

model_K = MultiPoolSCERMamba(
    temporal_features=13,
    static_features=12,
    hidden_size=64,
    num_layers=2,
    dropout=0.20
).to(device)

criterion_K = FocalLoss(
    gamma=2.0
)

optimizer_K = torch.optim.AdamW(
    model_K.parameters(),
    lr=0.0003,
    weight_decay=0.0001
)

MAX_EPOCHS = 20
PATIENCE = 4

best_pr = -1
best_roc = -1
best_f1 = -1
best_threshold = None

no_improvement = 0

BEST_PATH_K = "/content/best_experiment_K_J24.pt"


# ============================================================
# TRAIN — J24 TRAIN + VALIDATION ONLY
# ============================================================

for epoch in range(MAX_EPOCHS):

    model_K.train()

    total_loss = 0.0

    for sequence, static, targets in train_loader_J24:

        sequence = sequence.to(device)
        static = static.to(device)
        targets = targets.to(device)

        optimizer_K.zero_grad()

        logits, _ = model_K(
            sequence,
            static
        )

        loss = criterion_K(
            logits,
            targets
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model_K.parameters(),
            1.0
        )

        optimizer_K.step()

        total_loss += loss.item()

    # ========================================================
    # VALIDATION
    # ========================================================

    model_K.eval()

    probs_all = []
    labels_all = []

    with torch.no_grad():

        for sequence, static, targets in val_loader_J24:

            sequence = sequence.to(device)
            static = static.to(device)

            logits, _ = model_K(
                sequence,
                static
            )

            probs = torch.sigmoid(logits)

            probs_all.extend(
                probs.cpu().numpy()
            )

            labels_all.extend(
                targets.numpy()
            )

    probs_all = np.asarray(probs_all)
    labels_all = np.asarray(labels_all)

    pr_auc = average_precision_score(
        labels_all,
        probs_all
    )

    roc_auc = roc_auc_score(
        labels_all,
        probs_all
    )

    precision, recall, thresholds = (
        precision_recall_curve(
            labels_all,
            probs_all
        )
    )

    f1_values = (
        2 * precision[:-1] * recall[:-1]
        /
        (
            precision[:-1]
            + recall[:-1]
            + 1e-10
        )
    )

    idx = np.argmax(f1_values)

    epoch_f1 = float(
        f1_values[idx]
    )

    epoch_threshold = float(
        thresholds[idx]
    )

    print(
        f"Epoch {epoch+1:02d} | "
        f"Loss {total_loss / len(train_loader_J24):.4f} | "
        f"PR-AUC {pr_auc:.4f} | "
        f"ROC {roc_auc:.4f} | "
        f"F1 {epoch_f1:.4f}"
    )

    if pr_auc > best_pr:

        best_pr = pr_auc
        best_roc = roc_auc
        best_f1 = epoch_f1
        best_threshold = epoch_threshold

        no_improvement = 0

        torch.save(
            model_K.state_dict(),
            BEST_PATH_K
        )

        print("✅ Best Experiment K model saved")

    else:

        no_improvement += 1

        print(
            f"No improvement: "
            f"{no_improvement}/{PATIENCE}"
        )

        if no_improvement >= PATIENCE:
            print("EARLY STOPPING")
            break


print("\n======================================")
print("EXPERIMENT K COMPLETE")
print("======================================")

print("Best validation PR-AUC :", best_pr)
print("ROC-AUC                :", best_roc)
print("F1                     :", best_f1)
print("Threshold              :", best_threshold)
print("Model                  :", BEST_PATH_K)
```

```python
import os
import zipfile
import random
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import TensorDataset, DataLoader
from transformers import MambaConfig, MambaModel

# ============================================================
# DEVICE
# ============================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Using device:", device)


# ============================================================
# UNZIP J24 IF NEEDED
# ============================================================

zip_path = "/content/sequential_enhanced_model_ready_age24.zip"
extract_path = "/content/sequential_enhanced_model_ready_age24"

if not os.path.exists(extract_path):

    if not os.path.exists(zip_path):
        raise FileNotFoundError(
            "Upload sequential_enhanced_model_ready_age24.zip first."
        )

    with zipfile.ZipFile(zip_path, "r") as z:
        z.extractall(extract_path)

    print("✅ J24 unzipped")


# ============================================================
# FIND FILE
# ============================================================

def find_j24_file(filename):

    for root, dirs, files in os.walk(extract_path):

        if filename in files:
            return os.path.join(root, filename)

    raise FileNotFoundError(filename)


# ============================================================
# LOAD TRAIN + VALIDATION
# ============================================================

X_seq_train_J24 = np.load(
    find_j24_file("X_sequence_train.npy")
)

X_static_train_J24 = np.load(
    find_j24_file("X_static_train.npy")
)

y_train_J24 = np.load(
    find_j24_file("y_train.npy")
)

X_seq_val_J24 = np.load(
    find_j24_file("X_sequence_val.npy")
)

X_static_val_J24 = np.load(
    find_j24_file("X_static_val.npy")
)

y_val_J24 = np.load(
    find_j24_file("y_val.npy")
)


train_dataset_J24 = TensorDataset(
    torch.tensor(X_seq_train_J24, dtype=torch.float32),
    torch.tensor(X_static_train_J24, dtype=torch.float32),
    torch.tensor(y_train_J24, dtype=torch.float32)
)

val_dataset_J24 = TensorDataset(
    torch.tensor(X_seq_val_J24, dtype=torch.float32),
    torch.tensor(X_static_val_J24, dtype=torch.float32),
    torch.tensor(y_val_J24, dtype=torch.float32)
)

val_loader_J24 = DataLoader(
    val_dataset_J24,
    batch_size=256,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)


# ============================================================
# FOCAL LOSS
# ============================================================

class FocalLoss(nn.Module):

    def __init__(self, gamma=2.0):
        super().__init__()
        self.gamma = gamma

    def forward(self, logits, targets):

        targets = targets.float()

        bce = F.binary_cross_entropy_with_logits(
            logits,
            targets,
            reduction="none"
        )

        probs = torch.sigmoid(logits)

        pt = (
            probs * targets
            +
            (1 - probs) * (1 - targets)
        )

        return (
            ((1 - pt) ** self.gamma) * bce
        ).mean()


# ============================================================
# EXPERIMENT K MODEL
# ============================================================

class MultiPoolSCERMamba(nn.Module):

    def __init__(
        self,
        temporal_features=13,
        static_features=12,
        hidden_size=64,
        num_layers=2,
        dropout=0.20
    ):
        super().__init__()

        self.static_encoder = nn.Sequential(
            nn.Linear(static_features, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, hidden_size),
            nn.ReLU()
        )

        self.temporal_projection = nn.Linear(
            temporal_features,
            hidden_size
        )

        config = MambaConfig(
            vocab_size=2,
            hidden_size=hidden_size,
            state_size=16,
            num_hidden_layers=num_layers,
            expand=2,
            conv_kernel=4,
            use_cache=False
        )

        self.mamba = MambaModel(config)

        self.output_norm = nn.LayerNorm(
            hidden_size
        )

        self.temporal_attention = nn.Linear(
            hidden_size,
            1
        )

        self.risk_event_attention = nn.Sequential(
            nn.Linear(3, 16),
            nn.ReLU(),
            nn.Linear(16, 1)
        )

        self.classifier = nn.Sequential(
            nn.Linear(hidden_size * 4, 128),
            nn.ReLU(),
            nn.Dropout(dropout),

            nn.Linear(128, 32),
            nn.ReLU(),
            nn.Dropout(dropout),

            nn.Linear(32, 1)
        )

    def forward(self, sequence, static):

        static_embedding = self.static_encoder(
            static
        )

        temporal = self.temporal_projection(
            sequence
        )

        mamba_output = self.mamba(
            inputs_embeds=temporal,
            use_cache=False
        ).last_hidden_state

        mamba_output = self.output_norm(
            mamba_output
        )

        temporal_scores = self.temporal_attention(
            mamba_output
        ).squeeze(-1)

        risk_events = sequence[:, :, 4:7]

        risk_scores = self.risk_event_attention(
            risk_events
        ).squeeze(-1)

        attention_weights = torch.softmax(
            temporal_scores + risk_scores,
            dim=1
        )

        attention_pool = torch.sum(
            mamba_output *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        mean_pool = mamba_output.mean(dim=1)

        final_state = mamba_output[:, -1, :]

        combined = torch.cat(
            [
                attention_pool,
                mean_pool,
                final_state,
                static_embedding
            ],
            dim=1
        )

        logits = self.classifier(
            combined
        ).squeeze(-1)

        return logits, attention_weights


print("\n✅ RECOVERY COMPLETE")
print("Train:", X_seq_train_J24.shape)
print("Validation:", X_seq_val_J24.shape)
print("Risky train:", int(y_train_J24.sum()))
print("Risky validation:", int(y_val_J24.sum()))
print("Device:", device)
```

```python
import random
import numpy as np
import pandas as pd
import torch

from torch.utils.data import DataLoader
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    precision_recall_curve
)


def run_K_seed(seed):

    print("\n======================================")
    print("EXPERIMENT K SEED:", seed)
    print("======================================")

    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    generator = torch.Generator()
    generator.manual_seed(seed)

    train_loader_seed = DataLoader(
        train_dataset_J24,
        batch_size=256,
        shuffle=True,
        generator=generator,
        num_workers=2,
        pin_memory=True
    )

    model = MultiPoolSCERMamba(
        temporal_features=13,
        static_features=12,
        hidden_size=64,
        num_layers=2,
        dropout=0.20
    ).to(device)

    criterion = FocalLoss(
        gamma=2.0
    )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=0.0003,
        weight_decay=0.0001
    )

    best_pr = -1
    best_roc = -1
    best_f1 = -1
    best_threshold = None

    patience = 4
    no_improvement = 0

    model_path = f"/content/K_seed_{seed}.pt"

    for epoch in range(20):

        model.train()

        total_loss = 0.0

        for sequence, static, targets in train_loader_seed:

            sequence = sequence.to(device)
            static = static.to(device)
            targets = targets.to(device)

            optimizer.zero_grad()

            logits, _ = model(
                sequence,
                static
            )

            loss = criterion(
                logits,
                targets
            )

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                1.0
            )

            optimizer.step()

            total_loss += loss.item()

        # -----------------------------
        # VALIDATION
        # -----------------------------

        model.eval()

        probs_all = []
        labels_all = []

        with torch.no_grad():

            for sequence, static, targets in val_loader_J24:

                sequence = sequence.to(device)
                static = static.to(device)

                logits, _ = model(
                    sequence,
                    static
                )

                probs = torch.sigmoid(logits)

                probs_all.extend(
                    probs.cpu().numpy()
                )

                labels_all.extend(
                    targets.numpy()
                )

        probs_all = np.asarray(probs_all)
        labels_all = np.asarray(labels_all)

        pr_auc = average_precision_score(
            labels_all,
            probs_all
        )

        roc_auc = roc_auc_score(
            labels_all,
            probs_all
        )

        precision, recall, thresholds = (
            precision_recall_curve(
                labels_all,
                probs_all
            )
        )

        f1_values = (
            2 * precision[:-1] * recall[:-1]
            /
            (
                precision[:-1]
                + recall[:-1]
                + 1e-10
            )
        )

        idx = np.argmax(f1_values)

        epoch_f1 = float(
            f1_values[idx]
        )

        threshold = float(
            thresholds[idx]
        )

        print(
            f"Epoch {epoch+1:02d} | "
            f"PR-AUC {pr_auc:.4f} | "
            f"ROC {roc_auc:.4f} | "
            f"F1 {epoch_f1:.4f}"
        )

        if pr_auc > best_pr:

            best_pr = pr_auc
            best_roc = roc_auc
            best_f1 = epoch_f1
            best_threshold = threshold

            no_improvement = 0

            torch.save(
                model.state_dict(),
                model_path
            )

        else:

            no_improvement += 1

            if no_improvement >= patience:
                print("Early stopping")
                break

    del model
    torch.cuda.empty_cache()

    return {
        "seed": seed,
        "pr_auc": best_pr,
        "roc_auc": best_roc,
        "f1": best_f1,
        "threshold": best_threshold,
        "model_path": model_path
    }


# ============================================================
# RUN 3 SEEDS
# ============================================================

results_K = []

for seed in [42, 123, 2026]:

    result = run_K_seed(seed)

    results_K.append(result)


df_K = pd.DataFrame(results_K)

print("\n======================================")
print("EXPERIMENT K REPRODUCIBILITY")
print("======================================")

print(df_K)

print("\nMEAN")
print(
    df_K[
        ["pr_auc", "roc_auc", "f1"]
    ].mean()
)

print("\nSTANDARD DEVIATION")
print(
    df_K[
        ["pr_auc", "roc_auc", "f1"]
    ].std()
)

df_K.to_csv(
    "/content/experiment_K_reproducibility.csv",
    index=False
)

print("\nSaved:")
print("/content/experiment_K_reproducibility.csv")
```
